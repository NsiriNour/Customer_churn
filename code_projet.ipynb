## Contexte et Justification du Sujet

### Pourquoi ce sujet ?

Dans un marché des telecommunications hautement concurrentiel, la retention client est devenue 
l'un des enjeux strategiques les plus critiques pour les operateurs. Selon les estimations du secteur, 
**acquerir un nouveau client coute en moyenne 5 a 7 fois plus cher que de fidéliser un client existant**. 
Face a ce constat, la prediction du desabonnement (churn) constitue un levier decisif pour preserver 
la rentabilite d'une entreprise.

Nous avons choisi ce sujet pour cette raison :

1. **La pertinence metier** : Le churn est un probleme concret et mesurable, directement lie a la performance 
financiere des entreprises. La capacite a anticiper le depart d'un client transforme une reaction curative 
(remplacer le client perdu) en action preventive (proposer une offre de retention ciiblee).


### Importance et Impact du Projet

Un modele de prediction du churn bien calibre permet a une entreprise de telecoms de :

- **Identifier en avance** les clients a risque avant qu'ils ne resilient leur abonnement
- **Personnaliser les offres de retention** en ciblant les leviers exacts qui poussent un client au depart 
(type de contrat, niveau de facturation, qualite du service internet)
- **Optimiser le budget marketing** en concentrant les ressources sur le segment le plus vulnerable
- **Quantifier le risque financier** lie au portefeuille client en temps reel

Ce projet represente donc une application directe du Data Mining a une problematique a fort impact economique, 
illustrant comment les donnees peuvent etre transformees en decisions strategiques.

````python

# ============================================================
# DÉTECTION DES VALEURS ABERRANTES — MÉTHODE IQR
# ============================================================

def detect_outliers_iqr(data, column):
    """Détecte les valeurs aberrantes par la méthode de l'intervalle interquartile (IQR)"""
    Q1, Q3 = data[column].quantile(0.25), data[column].quantile(0.75)
    IQR = Q3 - Q1
    lb, ub = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    return data[(data[column] < lb) | (data[column] > ub)], lb, ub

print(f"{'Colonne':<20} {'Aberrantes':>10} {'Limite inf.':>15} {'Limite sup.':>15}")
print("-" * 62)
for col in num_cols:
    out, lb, ub = detect_outliers_iqr(df_raw.dropna(subset=[col]), col)
    print(f"{col:<20} {len(out):>10} {lb:>15.2f} {ub:>15.2f}")

print("""
Décision : Aucune suppression de valeurs aberrantes
  - tenure et MonthlyCharges : zéro valeur aberrante détectée
  - TotalCharges : quelques valeurs extrêmes légitimes (clients de haut profil)
  - Les modèles basés sur les arbres sont robustes aux aberrantes
  - StandardScaler dans le pipeline protège la Régression Logistique / SVM
""")
````

# 3. Preprocessing et Feature Engineering

Cette etape transforme les donnees brutes en un dataset propre et enrichi, pret pour la modelisation. 
Elle comprend trois phases : le nettoyage, la creation de nouvelles variables (feature engineering) 
et l'encodage des variables categorielles.


## 3.1 Nettoyage des Donnees


````python

# ============================================================
# PRÉTRAITEMENT DES DONNÉES
# ============================================================

df = df_raw.copy()

#  Suppression de l'ID client 
df.drop(columns=['customerID'], inplace=True)

#  Correction de TotalCharges (conversion en numérique et suppression des NaN) 
df.dropna(subset=['TotalCharges'], inplace=True)

print("Valeurs manquantes restantes :", df.isnull().sum().sum())

#  Encodage de la variable cible (Churn) 
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

print(df.shape)
print(df.dtypes)
````

## 3.2 Feature Engineering



L'objectif de cette étape est de créer de nouveaux indicateurs plus
représentatifs du comportement client pour améliorer la performance du modèle
de prédiction :

1.  AvgMonthlySpend (Intensité de dépense) : Calcul du ratio entre les charges
    mensuelles et l'ancienneté. Cet indicateur permet de détecter les clients
    qui paient cher proportionnellement à leur fidélité (segment à haut risque).
2.  NumServices (Score d'engagement) : Somme du nombre de services optionnels
    souscrits (Sécurité, Support, Streaming, etc.). Plus ce score est élevé,
    plus le client est "ancré" chez l'opérateur et moins il est susceptible de
    partir.
3.  HasInternet (Indicateur binaire) : Simplification de la variable
    InternetService en un indicateur de présence ou d'absence de connexion,
    permettant au modèle de se focaliser sur l'impact global de la connectivité.

Résultat : Ces nouvelles variables transforment des données brutes en leviers
décisionnels qui capturent mieux la fidélité et l'engagement du client.



````python

# ============================================================
# INGÉNIERIE DES FEATURES (FEATURE ENGINEERING)
# ============================================================

#  Feature 1 : Dépense mensuelle moyenne par ancienneté 
df['AvgMonthlySpend'] = df['MonthlyCharges'] / (df['tenure'] + 1)

#  Feature 2 : Nombre de services add-on souscrits 
addon_services = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
                  'TechSupport', 'StreamingTV', 'StreamingMovies']
df['NumServices'] = df[addon_services].apply(lambda row: (row == 'Yes').sum(), axis=1)

#  Feature 3 : Indicateur binaire d'avoir un service Internet 
df['HasInternet'] = (df['InternetService'] != 'No').astype(int)

print("Dimensions après ingénierie :", df.shape)
````

## 3.3 Encodage des Variables

###  Encodage des données (Conversion texte en chiffres)

On transforme les catégories en nombres pour que le modèle puisse les traiter :

*   **Label Encoding (Variables Binaires) :** Transforme les colonnes à 2 choix (ex: Oui/Non) en **0 et 1**. C'est simple et efficace.
*   **One-Hot Encoding (Multi-catégories) :** Pour les colonnes à 3 choix ou plus (ex: Type de Contrat), on crée des colonnes séparées. Cela évite au modèle de croire qu'une catégorie est "supérieure" à une autre.

**Résultat :** Le dataset devient 100% numérique, format indispensable pour le Machine Learning.

Le test ANOVA a été utilisé pour classer l'importance de nos variables numériques par rapport au **Churn**.

###  Comprendre les Statistiques : F-Score vs P-Value

Pour chaque variable, le test ANOVA génère deux indicateurs complémentaires :

#### 1. Le F-Score (La "Force")
*   **C'est quoi :** Il mesure **l'intensité** du signal.
*   **Logique :** Plus le F-Score est **élevé**, plus la variable permet de séparer nettement les clients qui partent de ceux qui restent. 
*   **Dans votre cas :** `AvgMonthlySpend` sépare beaucoup mieux les groupes que `NumServices`.

#### 2. La P-Value (La "Certitude")
*   **C'est quoi :** Elle mesure la **probabilité que le résultat soit dû au hasard**.
*   **Logique :** On cherche une P-Value la plus **petite** possible (généralement inférieure à **0.05**). 
    *   Si $P < 0.05$ : Le résultat est statistiquement significatif (on a confiance).
    *   Si $P > 0.05$ : Le résultat n'est pas fiable, même si le F-score semble élevé.
*   **Dans votre cas :** Pour ces variables, vos P-values sont extrêmement proches de **0**, ce qui signifie que ces relations sont réelles et très fiables.




````python

from sklearn.feature_selection import SelectKBest, f_classif

#  Définition des features numériques (y compris ceux créées) 
num_features = ['tenure', 'MonthlyCharges', 'TotalCharges', 'AvgMonthlySpend', 'NumServices']
X_num = df[num_features]
y = df['Churn']

#  Application du test ANOVA F sur les features numériques 
selector_anova = SelectKBest(score_func=f_classif, k='all')
selector_anova.fit(X_num, y)

#  Création d'un DataFrame pour les résultats 
anova_df = pd.DataFrame({
    'Feature': num_features,
    'F_Score': selector_anova.scores_,
    'P_Value': selector_anova.pvalues_
}).sort_values(by='F_Score', ascending=False)

#  Visualisation des F-scores 
plt.figure(figsize=(10, 5))
sns.barplot(data=anova_df, x='F_Score', y='Feature', palette='Blues_r')
plt.title("F-scores ANOVA : Features Numériques vs Churn")
plt.xlabel("F-Score (Plus élevé = Plus important)")
plt.show()

print("Classement ANOVA :")
print(anova_df)
````


###  Sélection des Variables Numériques (Test ANOVA)

Le classement par **F-Score** montre que :
1. Notre variable créée **`AvgMonthlySpend`** est le prédicteur le plus robuste du Churn.
2. La **`tenure`** (ancienneté) reste un facteur déterminant.
3. Toutes les variables testées présentent une **P-Value < 0.05**, confirmant que leur impact est statistiquement significatif et non dû au hasard.

**Décision :** Nous conserverons prioritairement les variables à haut F-Score pour optimiser la précision du modèle.

## 4.3 Elimination Recursive de Features (RFE)


````python

from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

#  Utilisation de toutes les features 
X = df.drop('Churn', axis=1)
y = df['Churn']

#  Initialisation du modèle et de RFE 
# Sélection de 12 features comme sous-ensemble cible
model = LogisticRegression(max_iter=1000, solver='liblinear')
rfe = RFE(estimator=model, n_features_to_select=12)
rfe.fit(X, y)

#  Création d'un DataFrame pour les résultats du classement RFE 
rfe_df = pd.DataFrame({
    'Feature': X.columns,
    'Rank': rfe.ranking_,
    'Selected': rfe.support_
}).sort_values(by='Rank')

#  Visualisation du classement RFE 
plt.figure(figsize=(10, 8))
rfe_plot_data = rfe_df.sort_values('Rank', ascending=True).head(20)
sns.barplot(data=rfe_plot_data, x=rfe_plot_data['Rank'].max() - rfe_plot_data['Rank'] + 1, 
            y='Feature', palette='mako')
plt.title("Classement RFE (Top 20 Features)")
plt.xlabel("Importance Relative (Plus élevé = meilleur)")
plt.show()

selected_rfe = rfe_df[rfe_df['Selected'] == True]['Feature'].tolist()
print("Features sélectionnées par RFE :", selected_rfe)
````

## 4.4 Elimination Recursive avec Validation Croisee (RFECV)


````python
from sklearn.feature_selection import RFECV
from sklearn.model_selection import StratifiedKFold
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Préparation des données
X = df.drop('Churn', axis=1)
y = df['Churn']

# 2. Initialisation du modèle de base
model = LogisticRegression(max_iter=1000, solver='liblinear')

# 3. Configuration de RFECV (avec Cross-Validation)
# cv=5 : découpe les données en 5 pour valider chaque test
# scoring='accuracy' : on cherche à optimiser la précision globale
rfecv = RFECV(estimator=model, step=1, cv=StratifiedKFold(5), scoring='accuracy')
rfecv.fit(X, y)

print(f"Nombre optimal de features : {rfecv.n_features_}")

# 4. Visualisation de la courbe de performance
plt.figure(figsize=(10, 6))
plt.title("Optimisation du nombre de variables (RFECV)")
plt.xlabel("Nombre de variables sélectionnées")
plt.ylabel("Score de précision (Cross-Validation)")

# On trace le score en fonction du nombre de variables
plt.plot(range(1, len(rfecv.cv_results_['mean_test_score']) + 1), 
         rfecv.cv_results_['mean_test_score'], color='blue', lw=2)

plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# 5. Liste des variables optimales retenues
best_features = X.columns[rfecv.support_].tolist()
print("Variables retenues :", best_features)
````

###  Note sur le choix de la méthode : RFE vs RFECV

*   **RFECV :** Aurait pu être utilisé pour trouver mathématiquement le nombre exact de variables nécessaires, mais au prix d'un temps de calcul plus élevé. 

**Décision :** Le RFE avec 12 variables offre ici le meilleur compromis entre performance prédictive et simplicité d'analyse pour les équipes marketing.

## 5.1 Modele 1 — Analyse Factorielle Discriminante (AFD / LDA)


L’**AFD (Analyse Factorielle Discriminante)** est une méthode statistique utilisée pour **séparer et classifier des groupes** à partir de plusieurs variables explicatives.

Dans un projet de **Customer Churn**, l’objectif de l’AFD est de distinguer :

- les clients qui vont quitter l’entreprise (**churners**),
- des clients qui vont rester.

L’AFD cherche des combinaisons linéaires des variables qui :

- maximisent la séparation entre les groupes,
- minimisent la variation à l’intérieur d’un même groupe.

````python
# 5.1  MODELE 1 : AFD / LDA (Analyse Discriminante Linéaire)
# Hypothèses : normalité + homogénéité des variances
# Robuste en pratique avec grand n (TCL)
# ============================================================
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import roc_auc_score, log_loss
import scipy.stats as ss
from scipy.stats import chi2 as chi2_dist

print('=' * 65)
print('5.1  AFD – Analyse Discriminante Linéaire (LDA)')
print('=' * 65)

# Utilisation de X_train_final (issu du preprocessing ColumnTransformer)
X_train_selected = X_train_final.values  # ou directement X_train_final
y_train_res = y_train  # adapter selon votre variable cible d'entraînement

lda = LinearDiscriminantAnalysis(solver='svd', store_covariance=True)
lda.fit(X_train_selected, y_train_res)

print(f'  Nb de fonctions discriminantes : {len(lda.explained_variance_ratio_)}')
print(f'  Variance expliquée DF1 : {lda.explained_variance_ratio_[0]*100:.2f}%')

# Coefficients discriminants
selected_features_rfe = X_train_final.columns.tolist()
coef_df = pd.DataFrame({
    'Variable'  : selected_features_rfe,
    'Coeff. LD1': lda.coef_[0]
}).sort_values('Coeff. LD1', key=abs, ascending=False)
print('\n  Coefficients discriminants (|coeff| décroissant) :')
print(coef_df.round(4).to_string(index=False))

acc_train = (lda.predict(X_train_selected) == y_train_res).mean()
print(f'\n  Taux de reclassement (train) : {acc_train*100:.2f}%')
````

Voici l'interprétation détaillée des **coefficients discriminants** de votre modèle. Ces chiffres sont les "poids" que le modèle donne à chaque variable pour décider si un client va rester ou partir.

### La règle de lecture

*   **Valeur Absolue** : Plus le chiffre est grand (loin de zéro), plus la variable est puissante.

###  Configuration du Modèle de Référence (Régression Logistique)

Pour établir notre base de performance (Baseline), nous utilisons une **Régression Logistique** avec les réglages standards suivants :

*   **Régularisation (C = 1.0) :** Ce paramètre contrôle l'équilibre entre l'ajustement aux données d'entraînement et la capacité du modèle à se généraliser à de nouveaux clients. La valeur 1.0 offre un compromis optimal : elle permet au modèle d'apprendre les motifs complexes du churn sans tomber dans le surapprentissage (*overfitting*).

### Gestion du Desequilibre des Classes — SMOTENC


````python

from imblearn.over_sampling import SMOTENC

#  1. Identification des indices des colonnes catégorielles 
# SMOTENC a besoin de connaître la position (0, 1, 2...) des features non-numériques
cat_indices = [i for i, col in enumerate(X_train_final.columns) 
               if col not in ['tenure', 'TotalCharges', 'AvgMonthlySpend']]

print(f"Indices des features catégorielles : {cat_indices}")

#  2. Application de SMOTENC aux données d'entraînement 
# On passe les indices au paramètre categorical_features
sm_nc = SMOTENC(categorical_features=cat_indices, random_state=42)
X_train_res, y_train_res = sm_nc.fit_resample(X_train_final, y_train)

#  3. Initialisation et entraînement du modèle avec SMOTE 
# Régression Logistique identique pour observer uniquement l'effet de SMOTENC
lr_smote = LogisticRegression(max_iter=1000, random_state=42)
lr_smote.fit(X_train_res, y_train_res)

#  4. Prédictions sur le jeu de test 
y_pred_smote = lr_smote.predict(X_test_final)
y_proba_smote = lr_smote.predict_proba(X_test_final)[:, 1]

print("\n" + "="*20 + " RÉGRESSION LOGISTIQUE : AVEC SMOTENC " + "="*20)
print(classification_report(y_test, y_pred_smote))

#  5. Tracé combiné ROC + Precision-Recall 
from sklearn.metrics import roc_curve, precision_recall_curve, auc, average_precision_score

fpr_smote, tpr_smote, _ = roc_curve(y_test, y_proba_smote)
precision_smote, recall_smote, _ = precision_recall_curve(y_test, y_proba_smote)
average_precision_smote = average_precision_score(y_test, y_proba_smote)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Courbes ROC et Precision-Recall pour la régression logistique avec SMOTENC', fontsize=14, fontweight='bold')

axes[0].plot(fpr_smote, tpr_smote, color='#1f77b4', lw=2, label=f'AUC = {auc(fpr_smote, tpr_smote):.3f}')
axes[0].plot([0, 1], [0, 1], color='gray', lw=1, linestyle='--', label='Aléatoire')
axes[0].set_title('Courbe ROC')
axes[0].set_xlabel('Taux de Faux Positifs (FPR)')
axes[0].set_ylabel('Taux de Vrais Positifs (TPR)')
axes[0].legend(loc='lower right')
axes[0].grid(alpha=0.3)

axes[1].plot(recall_smote, precision_smote, color='#d62728', lw=2, label=f'AP = {average_precision_smote:.3f}')
axes[1].set_title('Courbe Precision-Rappel')
axes[1].set_xlabel('Rappel')
axes[1].set_ylabel('Précision')
axes[1].legend(loc='lower left')
axes[1].grid(alpha=0.3)

plt.tight_layout(rect=[0, 0, 1, 0.96])
#  6. Métriques supplémentaires 
y_pred_smote = lr_smote.predict(X_test_final)
y_proba_smote = lr_smote.predict_proba(X_test_final)[:, 1]

from sklearn.metrics import confusion_matrix, roc_auc_score

cm_smote = confusion_matrix(y_test, y_pred_smote)
tn, fp, fn, tp = cm_smote.ravel()

sensitivity_smote = tp / (tp + fn)
specificity_smote = tn / (tn + fp)
auc_smote = roc_auc_score(y_test, y_proba_smote)

print("="*25 + " MÉTRIQUES SUPPLÉMENTAIRES AVEC SMOTENC " + "="*25)
print(f"AUC-ROC : {auc_smote:.4f}")
print(f"Sensibilité : {sensitivity_smote:.4f}")
print(f"Spécificité : {specificity_smote:.4f}")
print()
print("="*25 + " MATRICE DE CONFUSION AVEC SMOTENC " + "="*25)
print(cm_smote)
print()

plt.show()
````

Elle représente le compromis entre :

- **TPR (True Positive Rate / Recall)** en axe Y  
- **FPR (False Positive Rate)** en axe X  

La courbe est construite en testant différents seuils de classification à partir des probabilités prédites, puis en traçant les points (FPR, TPR).

## 5.3 Modele 3 — Arbre de Decision


## 🌳 Arbre de décision (projet Customer Churn)

Dans notre projet de **Customer Churn**, l’arbre de décision est un modèle de classification qui permet de prédire si un client va **quitter l’entreprise ou rester** en apprenant des règles simples à partir des variables (ex : ancienneté, consommation, type de contrat).

Le modèle construit une suite de décisions du type *“si… alors…”* afin de séparer progressivement les clients en groupes homogènes (churn / non churn).

## 📌 Paramètres utilisés (cas churn)
- **max_depth** : limite la profondeur pour éviter un modèle trop complexe (overfitting).
- **min_samples_split** : nombre minimum de clients pour créer une séparation.
- **min_samples_leaf** : taille minimale d’un groupe final de clients.
- **criterion (gini / entropy)** : mesure de qualité de séparation entre churn et non churn.



L'élagage simplifie l'arbre en supprimant les branches trop spécifiques pour éviter que le modèle n'apprenne les données par cœur (overfitting). Grâce à l'alpha optimal, nous obtenons un modèle robuste de 15 feuilles qui capture les tendances majeures du churn tout en restant parfaitement interprétable.
pur cela on a fait le tableau CP pour trouver l'alpha optimal 

````python
# ============================================================
# ARBRE DE DÉCISION — Table CP et sélection de best_alpha
# ============================================================
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier
#  Création et entraînement du modèle d'arbre de décision 
tree_model = DecisionTreeClassifier()
#  Calcul du chemin de complexité-coût de l'élagage 
path = tree_model.cost_complexity_pruning_path(X_train_final, y_train)
ccp_alphas, impurities = path.ccp_alphas, path.impurities  # ← fix
#  Évaluation des F-scores pour différentes valeurs d'alpha 
cp_scores = []
for alpha in ccp_alphas[:-1]:
    clf = DecisionTreeClassifier(random_state=42, ccp_alpha=alpha)
    scores = cross_val_score(clf, X_train_final, y_train, cv=5, scoring='roc_auc', n_jobs=-1)
    cp_scores.append((alpha, scores.mean(), scores.std()))
cp_scores_df = pd.DataFrame(cp_scores, columns=['ccp_alpha', 'AUC_mean', 'AUC_std'])
cp_scores_df = cp_scores_df.sort_values('ccp_alpha')
print("\n  ccp_alpha    AUC_mean    AUC_std")
print(cp_scores_df[['ccp_alpha', 'AUC_mean', 'AUC_std']].head(10).to_string(index=True, float_format='%.6f'))
#  Identification de l'alpha optimal 
best_alpha = float(cp_scores_df.loc[cp_scores_df['AUC_mean'].idxmax(), 'ccp_alpha'])
print(f"\n  Alpha optimal : {best_alpha:.6f}")
#  Entraînement de l'arbre élagué 
best_tree = DecisionTreeClassifier(random_state=42, ccp_alpha=best_alpha)
best_tree.fit(X_train_final, y_train)
print(f"  Arbre élagué : {best_tree.get_n_leaves()} feuilles, profondeur {best_tree.get_depth()}")
````

````python

# ============================================================
# ÉVALUATION ARBRE DE DÉCISION — métriques, ROC, PR et matrice de confusion
# ============================================================
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, precision_recall_curve, average_precision_score

#  Prédictions et probabilités 
ny_pred_tree = best_tree.predict(X_test_final)
y_proba_tree = best_tree.predict_proba(X_test_final)[:, 1]

print("Rapport de classification - Arbre de Décision :")
print(classification_report(y_test, ny_pred_tree, target_names=['Pas de Churn', 'Churn']))

#  Matrice de confusion 
cm_tree = confusion_matrix(y_test, ny_pred_tree)
print("Matrice de confusion :")
print(cm_tree)

#  Calcul des courbes ROC et Precision-Recall 
roc_fpr, roc_tpr, _ = roc_curve(y_test, y_proba_tree)
roc_auc_tree = auc(roc_fpr, roc_tpr)
precision_tree, recall_tree, _ = precision_recall_curve(y_test, y_proba_tree)
average_precision_tree = average_precision_score(y_test, y_proba_tree)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Arbre de Décision - Courbes ROC et Precision-Recall', fontsize=14, fontweight='bold')

#  Tracé de la courbe ROC 
axes[0].plot(roc_fpr, roc_tpr, color='#1f77b4', lw=2, label=f'AUC = {roc_auc_tree:.3f}')
axes[0].plot([0,1], [0,1], color='gray', ls='--')
axes[0].set_title('Courbe ROC')
axes[0].set_xlabel('FPR')
axes[0].set_ylabel('TPR')
axes[0].legend(loc='lower right')
axes[0].grid(alpha=0.3)

#  Tracé de la courbe Precision-Recall 
axes[1].plot(recall_tree, precision_tree, color='#d62728', lw=2, label=f'AP = {average_precision_tree:.3f}')
axes[1].set_title('Courbe Precision-Rappel')
axes[1].set_xlabel('Rappel')
axes[1].set_ylabel('Précision')
axes[1].legend(loc='lower left')
axes[1].grid(alpha=0.3)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
````

L'arbre de décision affiche une performance globale solide avec une exactitude de **77%** et un **AUC de 0,815**, démontrant une bonne capacité à distinguer les deux types de clients.

Cependant, son point faible reste le **rappel pour le Churn (0,50)** : le modèle ne détecte qu'un client sur deux qui part réellement, ce qui est inférieur aux performances de la régression logistique. Bien que cet arbre soit très simple à interpréter visuellement, il s'avère moins efficace pour une stratégie de rétention massive.

## 5.4 Modele 4 — Random Forest


le **Random Forest** est un modèle de classification basé sur plusieurs arbres de décision permettant de prédire si un client va quitter l’entreprise ou rester.

Le modèle fonctionne grâce au principe du **bootstrap** :
pour chaque arbre, un nouvel échantillon aléatoire des données est créé avec remise. Cela signifie qu’une même observation peut être sélectionnée plusieurs fois tandis que d’autres peuvent ne pas être choisies. Chaque arbre est donc entraîné sur des données légèrement différentes.

Ensuite :
- chaque arbre effectue sa propre prédiction,
- puis la prédiction finale est obtenue par **vote majoritaire** entre tous les arbres.

Cette diversité entre les arbres permet :
- de réduire l’overfitting,
- d’améliorer la robustesse,
- et d’obtenir une meilleure généralisation sur de nouvelles données.

###  Paramètres principaux
- **n_estimators** : nombre d’arbres dans la forêt.
- **max_depth** : profondeur maximale des arbres.
- **min_samples_split** : minimum d’observations pour effectuer une séparation.
- **min_samples_leaf** : minimum d’observations dans une feuille.
- **max_features** : nombre de variables testées à chaque division.
- **bootstrap** : active l’échantillonnage aléatoire avec remise.

### Résumé
Le Random Forest combine plusieurs arbres de décision entraînés sur des échantillons bootstrap différents afin d’améliorer la précision et la stabilité de la prédiction des clients à risque de churn.

````python

# ============================================================
# MODÈLE RANDOM FOREST — création et paramètres importants
# ============================================================
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

#  Création et entraînement du modèle Random Forest 
rf_model = RandomForestClassifier(random_state=42, n_estimators=100, max_depth=None)
rf_model.fit(X_train_final, y_train)

print("Paramètres importants du modèle Random Forest :")
important_params = ['n_estimators', 'max_depth', 'min_samples_split', 'min_samples_leaf', 'max_features', 'bootstrap', 'criterion', 'random_state']
for param in important_params:
    print(f"{param}: {rf_model.get_params()[param]}")

print("\nTop 10 des importances des features :")
feature_importances = pd.DataFrame({
    'feature': X_train_final.columns,
    'importance': rf_model.feature_importances_
}).sort_values('importance', ascending=False)
print(feature_importances.head(10).to_string(index=False))
````

Le Random Forest utilise un ensemble de 100 arbres (n_estimators) entraînés sur des échantillons aléatoires (bootstrap) et des sous-ensembles de variables (max_features). 
Cette stratégie de diversité permet de compenser la tendance des arbres individuels au surapprentissage, créant ainsi un modèle de prédiction du churn beaucoup plus stable et robuste qu'un arbre unique
L'importance calculée par un Random Forest est spécifique au Random Forest. Une variable jugée "importante" par la forêt ne sera pas forcément la meilleure pour une Régression Logistique.

````python

# ============================================================
# ÉVALUATION RANDOM FOREST — métriques, spécificité, sensibilité, AUC
# ============================================================
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, precision_recall_curve, average_precision_score
import matplotlib.pyplot as plt

#  Prédictions 
y_pred_rf = rf_model.predict(X_test_final)
y_proba_rf = rf_model.predict_proba(X_test_final)[:, 1]

#  Rapport de classification 
print("Rapport de classification - Random Forest :")
print(classification_report(y_test, y_pred_rf, target_names=['Pas de Churn', 'Churn']))

#  Matrice de confusion 
cm_rf = confusion_matrix(y_test, y_pred_rf)
print("\nMatrice de confusion :")
print(cm_rf)

#  Calcul de la sensibilité et spécificité 
tn, fp, fn, tp = cm_rf.ravel()
sensitivity = tp / (tp + fn)  #  Sensibilité = Rappel = TP/(TP+FN) 
specificity = tn / (tn + fp)  #  Spécificité = TN/(TN+FP) 

print(f"\nSensibilité (Rappel) : {sensitivity:.3f}")
print(f"Spécificité : {specificity:.3f}")

#  Calcul de l'AUC 
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_proba_rf)
roc_auc_rf = auc(fpr_rf, tpr_rf)
print(f"AUC : {roc_auc_rf:.3f}")

#  Courbes ROC et Precision-Recall 
precision_rf, recall_rf, _ = precision_recall_curve(y_test, y_proba_rf)
average_precision_rf = average_precision_score(y_test, y_proba_rf)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Random Forest - Courbes ROC et Precision-Rappel', fontsize=14, fontweight='bold')

#  Courbe ROC 
axes[0].plot(fpr_rf, tpr_rf, color='#1f77b4', lw=2, label=f'AUC = {roc_auc_rf:.3f}')
axes[0].plot([0, 1], [0, 1], color='gray', ls='--')
axes[0].set_title('Courbe ROC')
axes[0].set_xlabel('Taux de Faux Positifs (FPR)')
axes[0].set_ylabel('Taux de Vrais Positifs (TPR)')
axes[0].legend(loc='lower right')
axes[0].grid(alpha=0.3)

#  Courbe Precision-Rappel 
axes[1].plot(recall_rf, precision_rf, color='#d62728', lw=2, label=f'AP = {average_precision_rf:.3f}')
axes[1].set_title('Courbe Precision-Rappel')
axes[1].set_xlabel('Rappel')
axes[1].set_ylabel('Précision')
axes[1].legend(loc='lower left')
axes[1].grid(alpha=0.3)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
````

````python
from sklearn.model_selection import RandomizedSearchCV

# ============================================================
# FINE-TUNING — XGBOOST (RandomizedSearchCV)
# ============================================================
print('=' * 65)
print('Fine-Tuning — XGBoost')
print('=' * 65)

param_xgb = {
    'n_estimators'    : [100, 200, 300],
    'max_depth'       : [3, 5, 7, 9],
    'learning_rate'   : [0.01, 0.05, 0.1, 0.2],
    'subsample'       : [0.6, 0.8, 1.0],
    'colsample_bytree': [0.6, 0.8, 1.0],
    'gamma'           : [0, 1, 3],
    'reg_alpha'       : [0, 0.1, 1],
    'reg_lambda'      : [1, 2, 4],
    'scale_pos_weight': [1, y_train.value_counts()[0] / y_train.value_counts()[1]],
}
grid_xgb = RandomizedSearchCV(
    XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='logloss', n_jobs=-1),
    param_xgb,
    n_iter=20,
    cv=StratifiedKFold(3),
    scoring='roc_auc',
    random_state=42,
    n_jobs=-1
)
grid_xgb.fit(X_train_res, y_train_res)

print(f'  Meilleurs paramètres : {grid_xgb.best_params_}')
print(f'  Meilleur AUC-CV      : {grid_xgb.best_score_:.4f}')
xgb_best   = grid_xgb.best_estimator_
y_prob_xgb = xgb_best.predict_proba(X_test_final)[:,1]
y_pred_xgb = xgb_best.predict(X_test_final)
print(f'  AUC-ROC (test)       : {roc_auc_score(y_test, y_prob_xgb):.4f}')

print('  Meilleurs paramètres XGBoost :')
for p, v in xgb_best.get_params().items():
    if p in ['n_estimators','max_depth','learning_rate','subsample','colsample_bytree','gamma','reg_alpha','reg_lambda','scale_pos_weight']:
        print(f'  {p:<20} = {v}')
````

Si la meilleure combinaison sélectionne un `learning_rate` bas et un `max_depth` modéré, cela signifie que le modèle évite le surapprentissage tout en conservant une forte capacité prédictive.
Le `scale_pos_weight` personnalisé permet de mieux compenser le déséquilibre des classes dans le churn.


La différence principale réside dans la **méthode de recherche** et la **vitesse**.

### 1. GridSearchCV (L'exhaustif)
*   **Principe :** Il teste **absolument toutes** les combinaisons possibles de la grille que vous avez définie.
*   **Avantage :** Il garantit de trouver la meilleure combinaison mathématique au sein de votre grille.
*   **Inconvénient :** Il est **très lent** si vous avez beaucoup de paramètres (si vous avez 100 combinaisons, il fera 100 entraînements).

### 2. RandomizedSearchCV (Le sélectif)
*   **Principe :** Il choisit des combinaisons **au hasard** dans la grille, et s'arrête après un nombre d'essais que vous fixez (ex: `n_iter=10`).
*   **Avantage :** Il est **beaucoup plus rapide**. On l'utilise souvent pour explorer de très grandes grilles (comme pour un Random Forest ou XGBoost).
*   **Inconvénient :** Il peut rater la "meilleure" combinaison absolue, mais il trouve généralement une solution très proche de l'optimale en un temps record.

### Interprétation — Optimisation : Random Forest (RandomizedSearchCV)

**Pourquoi RandomizedSearch ?**
Avec 5 hyperparamètres et plusieurs valeurs chacun, l'exploration exhaustive nécessiterait
162 × 3 folds = 486 entraînements de RF — trop lent. RandomizedSearchCV tire aléatoirement
12 combinaisons, couvrant mieux l'espace de recherche en une fraction du temps.

- `n_estimators` optimal : à partir de combien d'arbres la stabilité est atteinte
- `max_features` : `'sqrt'` vs `'log2'` contrôlent la décorrélation entre arbres
- `class_weight = 'balanced_subsample'` : recalcule les poids à chaque bootstrap —
  plus robuste que `'balanced'` global pour les Random Forests sur données déséquilibrées

## 7.3 Courbes d'Apprentissage — Diagnostic Biais/Variance


````python
# ============================================================
# COURBES D'APPRENTISSAGE (Diagnostic Biais/Variance) — 5 modèles
# ============================================================
from sklearn.model_selection import learning_curve
print('=' * 65)
print("COURBES D'APPRENTISSAGE")
print('=' * 65)
fig, axes = plt.subplots(1, 5, figsize=(25,5), sharey=True)
mlist = [
    (lda_best,   X_train_final, 'AFD (LDA)',              COLORS[0]),
    (logit_best, X_train_final, 'Régression Logistique',  COLORS[1]),
    (tree_best,  X_train_final, 'Arbre de Décision',      COLORS[2]),
    (rf_best,    X_train_final, 'Random Forest',          COLORS[3]),
    (xgb_best,   X_train_final, 'XGBoost',                COLORS[4]),
]
for ax, (model, X_tr, name, color) in zip(axes, mlist):
    y_tr = y_train
    #  Calcul des courbes d'apprentissage 
    tr_s, tr_sc, val_sc = learning_curve(
        model, X_tr, y_tr,
        cv=StratifiedKFold(5), scoring='roc_auc',
        train_sizes=np.linspace(0.1,1.0,6), n_jobs=-1)
    #  Tracé de l'erreur d'entraînement 
    ax.plot(tr_s, tr_sc.mean(1), 'o-', color=color, label='Entraînement')
    #  Tracé de l'erreur de validation 
    ax.plot(tr_s, val_sc.mean(1), 's--', color=color, alpha=0.7, label='Validation CV')
    #  Zone de confiance autour de la validation 
    ax.fill_between(tr_s, val_sc.mean(1)-val_sc.std(1),
                    val_sc.mean(1)+val_sc.std(1), alpha=0.1, color=color)
    ax.set_title(name, fontweight='bold', fontsize=9)
    ax.set_xlabel('Taille entraînement'); ax.set_ylabel('AUC-ROC')
    ax.legend(fontsize=8); ax.set_ylim([0.5,1.02])
plt.suptitle("Courbes d'Apprentissage — Diagnostic Biais/Variance (5 modèles)",
             fontsize=13, fontweight='bold')
plt.tight_layout(); plt.show()
````

### Interprétation — Courbes d'Apprentissage (5 Modèles)

- **Sur-apprentissage** : grand écart train vs CV → le modèle mémorise (solution : régularisation)
- **Sous-apprentissage** : train ET CV bas → capacité insuffisante
- **Bonne généralisation** : train et CV convergent à une valeur élevée

**Observations par modèle :**
- **AFD (LDA)** : Convergence rapide et stable — léger biais structurel dû à l'hypothèse de linéarité.
- **Régression Logistique** : Courbes train/CV très proches → excellent équilibre biais/variance, le modèle généralise bien.
- **Arbre de Décision** : Léger sur-apprentissage visible (train > CV) même après élagage — la profondeur reste un facteur de risque.
- **Random Forest** : Sur-apprentissage modéré (train supérieur à CV) typique des forêts, mais le gap se réduit avec l'optimisation.
- **XGBoost** : Convergence progressive et stable grâce au `learning_rate` faible. Si train ≈ CV → excellente généralisation ; si gap > 0.05 → augmenter `reg_lambda` ou réduire `max_depth`.

**Conclusion :** XGBoost et la Régression Logistique présentent les meilleures courbes d'apprentissage pour ce problème de churn, confirmant leur robustesse sur des données réelles déséquilibrées.


## Conclusion générale

# 8. Interpretabilite — SHAP et Importance des Variables (XAI)

---



Le **SHAP** (SHapley Additive exPlanations) est l'outil le plus puissant actuellement en **IA Explicable (XAI)**. 

Son but est simple : **Ouvrir la "boîte noire" du modèle** pour comprendre exactement pourquoi il a pris une décision pour un client précis.

### 2. Comment ça marche concrètement ?
Le SHAP compare la prédiction du modèle avec et sans une variable, dans toutes les combinaisons possibles. 
*   **Valeur SHAP Positive (+)** : La variable a poussé le client vers le **Churn** (le départ).
*   **Valeur SHAP Négative (-)** : La variable a poussé le client vers la **Fidélité** (il reste).

### 3. Pourquoi c'est mieux que la "Feature Importance" classique ?
La "Feature Importance" classique vous dit : *"Le Contrat est important"*. C'est tout.
**Le SHAP vous dit beaucoup plus :**
1.  **Le sens :** Est-ce que le contrat aide à rester ou pousse à partir ?
2.  **L'impact :** De combien de % cette variable a-t-elle modifié la probabilité finale ?
3.  **L'explication locale :** Vous pouvez prendre le **Client n°452** et dire : *"Lui, il est prédit partant à 80% à cause de sa facture élevée, même si son contrat de 1 an essaie de le retenir."*

---


## 8.1 Feature Importance — MDI vs Permutation


````python

# ============================================================
# FEATURE IMPORTANCE : MDI vs Permutation
# ============================================================
from sklearn.inspection import permutation_importance

print('=' * 65)
print('FEATURE IMPORTANCE — MDI vs Permutation')
print('=' * 65)

#  Calcul de MDI (Mean Decrease Impurity) 
fi_mdi  = pd.DataFrame({'Variable': X_train_res.columns,
                         'MDI': rf_best.feature_importances_})
#  Calcul de l'importance par permutation 
perm    = permutation_importance(rf_best, X_test_final, y_test,
                                  n_repeats=10, random_state=42,
                                  scoring='roc_auc', n_jobs=-1)
fi_perm = pd.DataFrame({'Variable' : X_test_final.columns,
                         'Perm_mean': perm.importances_mean,
                         'Perm_std' : perm.importances_std})
fi_cmp  = fi_mdi.merge(fi_perm, on='Variable').sort_values('MDI', ascending=False)
print(fi_cmp.round(4).to_string(index=False))

#  Visualisation comparée des deux méthodes 
fig, axes = plt.subplots(1, 2, figsize=(18,7))
fi_s1 = fi_cmp.sort_values('MDI', ascending=True)
axes[0].barh(fi_s1['Variable'], fi_s1['MDI'], color='#FF9800')
axes[0].set_title('MDI — Diminution Moyenne d\'Impureté (RF)', fontweight='bold')
axes[0].set_xlabel('Score MDI')

fi_s2 = fi_cmp.sort_values('Perm_mean', ascending=True)
axes[1].barh(fi_s2['Variable'], fi_s2['Perm_mean'],
             xerr=fi_s2['Perm_std'], color='#2196F3', capsize=3)
axes[1].set_title('Importance par Permutation — RF (Jeu de Test)', fontweight='bold')
axes[1].set_xlabel('Réduction AUC-ROC moyenne')
plt.tight_layout(); plt.show()

````

### Interprétation — Feature Importance MDI vs Permutation

**MDI (Mean Decrease Impurity) :**
Calculé pendant l'entraînement — somme des réductions de Gini attribuées à chaque
variable sur tous les arbres. Rapide mais biaisé vers les variables à haute cardinalité.

**Permutation Importance :**
Calculé sur le jeu de **test** — mesure la chute d'AUC après permutation aléatoire
de chaque variable. Plus fiable car évalue l'importance réelle pour la généralisation.

**Consensus :** les variables systématiquement en tête des deux classements constituent
les vrais drivers du churn : `tenure`, `Contract_Two year`, `TotalCharges`,
`InternetService_Fiber optic`.



### 2. Les grands gagnants
*   **`AvgMonthlySpend` :** C'est la variable **n°1**. Elle a un score élevé dans les deux colonnes. Cela confirme que votre "feature engineering" a créé la variable la plus utile pour prédire le churn.
*   **`Contract_Two year` :** Elle est très importante en Permutation (**0.0444**). Cela signifie que le type de contrat est un pilier fondamental de la décision du modèle.

### 3. Le paradoxe de `TotalCharges` et `tenure`
*   Regardez bien : elles ont un score **MDI élevé**, mais un score de **Permutation proche de zéro ou négatif**.
*   **Interprétation :** Le Random Forest les utilise par "facilité" parce que ce sont des chiffres, mais si on les enlève, le modèle ne perd pas en précision car l'information est déjà présente dans `AvgMonthlySpend`. Elles sont donc **redondantes**.
